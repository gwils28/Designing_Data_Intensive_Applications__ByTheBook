```python
import sqlite3
import time

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from ddia import requirements as rq
from ddia.diagrams import Diagram
from ddia.plotting import (AMBER, GRAY, GREEN, INDIGO, LIGHT, ROSE, SKY, SLATE, TEAL, VIOLET, COLORS,
                           annotate_bars, finish, set_style, subtitle)

set_style()
plt.rcParams["axes.titlepad"] = 16
pd.set_option("display.precision", 3)
rng = np.random.default_rng(0)
fmt = lambda v: f"{v:,.0f}".replace(",", " ")      # 1234567 -> « 1 234 567 »
```

## 1. Le schéma relationnel et la requête de timeline (section *Representing Users, Posts, and Follows*)

**Le problème.** Chaque utilisateur doit voir sa **timeline d'accueil** : les messages
récents des comptes qu'il suit, du plus récent au plus ancien. Le modèle le plus direct
tient en trois tables : `users`, `posts` (avec l'identifiant de l'auteur et un
horodatage) et `follows` (une ligne par relation « A suit B »). La timeline est alors une
**jointure** : trouver les comptes suivis, leurs messages récents, trier et garder les
premiers.

Construisons ce schéma sur SQLite avec un graphe social synthétique de 5 000 comptes
(fonction `rq.social_graph` : la popularité suit une loi de puissance, nous y
reviendrons en section 4).

```python
N_USERS = 5_000
follower, followee = rq.social_graph(N_USERS, mean_follows=50, alpha=1.2, seed=1)
n_followers, n_following = rq.degree_counts(follower, followee, N_USERS)

con = sqlite3.connect(":memory:")
con.executescript("""
CREATE TABLE users   (id INTEGER PRIMARY KEY, name TEXT);
CREATE TABLE posts   (id INTEGER PRIMARY KEY, sender_id INTEGER, timestamp REAL, text TEXT);
CREATE TABLE follows (follower_id INTEGER, followee_id INTEGER, PRIMARY KEY (follower_id, followee_id));
CREATE INDEX posts_by_sender ON posts (sender_id, timestamp);
""")
con.executemany("INSERT INTO users VALUES (?, ?)", ((i, f"user{i}") for i in range(N_USERS)))
con.executemany("INSERT INTO follows VALUES (?, ?)", zip(follower.tolist(), followee.tolist()))

# 40 messages par compte en moyenne, sur 30 jours, auteurs tirés uniformément
N_POSTS = 200_000
sender = rng.integers(0, N_USERS, N_POSTS)
ts = np.sort(rng.uniform(0, 30 * 86400, N_POSTS))
con.executemany("INSERT INTO posts VALUES (?, ?, ?, ?)",
                ((i, int(s), float(t), f"message {i}") for i, (s, t) in enumerate(zip(sender, ts))))
con.commit()
pd.DataFrame({"table": ["users", "posts", "follows"],
              "lignes": [con.execute(f"SELECT COUNT(*) FROM {t}").fetchone()[0] for t in ["users", "posts", "follows"]]})
```

La requête de timeline, adaptée de celle du livre (avec une limite de 50 messages au
lieu de 1 000 pour notre petit jeu de données) :

```python
TIMELINE_SQL = """
SELECT posts.id, posts.sender_id, posts.timestamp, users.name
  FROM posts
  JOIN follows ON posts.sender_id = follows.followee_id
  JOIN users   ON posts.sender_id = users.id
 WHERE follows.follower_id = ?
 ORDER BY posts.timestamp DESC
 LIMIT 50
"""
for row in con.execute("EXPLAIN QUERY PLAN " + TIMELINE_SQL, (0,)):
    print(row[-1])
```

Le plan de requête confirme l'intuition du livre : SQLite parcourt la liste des comptes
suivis (clé primaire de `follows`), va chercher pour chacun ses messages via l'index
`posts_by_sender`, puis **trie** l'ensemble dans une structure temporaire (`USE TEMP
B-TREE FOR ORDER BY`). Le travail est donc proportionnel au nombre de comptes suivis et à
leur activité, et non à la taille de la timeline demandée. Mesurons le temps de la
requête pour des utilisateurs qui suivent de quelques comptes à plusieurs centaines.

```python
def time_query(sql, arg, repeat=5):
    best = np.inf
    for _ in range(repeat):
        t0 = time.perf_counter()
        con.execute(sql, (arg,)).fetchall()
        best = min(best, time.perf_counter() - t0)
    return best * 1e3

sample_users = rng.choice(N_USERS, 300, replace=False)
mes = pd.DataFrame({"suivis": n_following[sample_users],
                    "ms": [time_query(TIMELINE_SQL, int(u)) for u in sample_users]})
coef = np.polyfit(mes.suivis, mes.ms, 1)

fig, ax = plt.subplots(figsize=(8.5, 4))
ax.scatter(mes.suivis, mes.ms, s=12, color=INDIGO, alpha=0.6, label="un utilisateur")
xs = np.linspace(0, mes.suivis.max(), 50)
ax.plot(xs, np.polyval(coef, xs), color=ROSE, label=f"ajustement linéaire : {coef[0] * 1e3:.2f} µs par compte suivi".replace(".", ","))
ax.set_xlabel("nombre de comptes suivis")
ax.set_ylabel("temps de la requête (ms)")
ax.set_title("Le coût de la timeline « à la lecture » croît avec le nombre de comptes suivis")
subtitle(ax, f"SQLite en mémoire, {fmt(N_POSTS)} messages, meilleur de 5 essais")
ax.legend()
plt.tight_layout(); plt.show()
```

Le coût est à peu près **linéaire** dans le nombre de comptes suivis : chaque compte
suivi ajoute une recherche dans l'index des messages et des lignes à trier. Sur une
machine et quelques milliers de comptes, c'est rapide. Le problème n'est pas une requête,
c'est leur **nombre** : passons aux ordres de grandeur.

> **À retenir.** La timeline relationnelle est une jointure suivie d'un tri. Son coût
> est proportionnel au nombre de comptes suivis : c'est un calcul **à la lecture**,
> refait à chaque consultation.

## 2. Ordres de grandeur de la charge

Le livre donne des chiffres de l'ordre de ceux d'un grand réseau social. Nous les
mettons dans un dictionnaire pour calculer les conséquences des deux stratégies.

Les hypothèses du livre : 500 millions de messages par jour (5 800 par seconde en moyenne,
avec des pointes à 150 000), 200 abonnés et 200 abonnements en moyenne. Pour la lecture,
le livre imagine d'abord que les clients **interrogent** le serveur toutes les 5 secondes
pour voir les nouveaux messages (polling) : avec 10 millions d'utilisateurs connectés en
même temps, cela fait 2 millions de requêtes de timeline par seconde.

```python
connectes, periode_polling_s = 10_000_000, 5
CHARGE = dict(publications_par_s=500e6 / 86400, pic_publications_par_s=150_000,
              lectures_timeline_par_s=connectes / periode_polling_s, abonnes_moyen=200)
```

Une remarque simple mais utile : dans un graphe fermé, **le nombre moyen d'abonnés est
égal au nombre moyen d'abonnements** (chaque arête est un abonnement pour l'un et un
abonné pour l'autre). On peut donc utiliser le même chiffre pour les deux.

- **À la lecture** (requête SQL) : chaque lecture de timeline consulte les messages de
  chaque compte suivi, soit `lectures × abonnements` accès ;
- **à l'écriture** (fan-out) : chaque message est copié dans la timeline de chaque
  abonné, soit `publications × abonnés` écritures, et une lecture ne coûte qu'un accès.

```python
c = CHARGE
couts = pd.DataFrame({
    "stratégie": ["à la lecture", "à la lecture", "à l'écriture", "à l'écriture"],
    "opération": ["écritures (messages)", "accès en lecture", "écritures (fan-out)", "accès en lecture"],
    "par seconde": [c["publications_par_s"], c["lectures_timeline_par_s"] * c["abonnes_moyen"],
                    c["publications_par_s"] * c["abonnes_moyen"], c["lectures_timeline_par_s"]],
})
fig, ax = plt.subplots(figsize=(9.5, 3.8))
labels = [f"{a}\n{b}" for a, b in zip(couts["stratégie"], couts["opération"])]
colors = [GRAY, INDIGO, AMBER, INDIGO]
ax.barh(labels[::-1], couts["par seconde"].values[::-1], color=colors[::-1], height=0.6)
ax.set_xscale("log"); rq.plain_log_axis(ax, "x")
for i, v in enumerate(couts["par seconde"].values[::-1]):
    ax.text(v * 1.15, i, fmt(v) + " /s", va="center", fontsize=9.5)
ax.set_xlim(1e3, couts["par seconde"].max() * 30)
ax.set_xlabel("opérations élémentaires par seconde (échelle log)")
ax.set_title("Recalculer à chaque lecture coûte des ordres de grandeur de plus")
subtitle(ax, f"{fmt(c['publications_par_s'])} messages/s ; {fmt(c['lectures_timeline_par_s'])} lectures de timeline/s ; "
             f"{c['abonnes_moyen']} abonnés en moyenne")
ax.grid(axis="y", visible=False)
plt.tight_layout(); plt.show()
couts.assign(**{"par seconde": couts["par seconde"].map(fmt)})
```

Avec ces hypothèses, le polling coûterait 400 millions d'accès par seconde, contre
environ 1,2 million d'écritures de fan-out (plus 2 millions de lectures de cache, et
même bien moins si le serveur **pousse** les nouveaux messages aux clients connectés au
lieu d'être interrogé toutes les 5 secondes). L'écart vient du **rapport
lectures/écritures** : on consulte sa timeline des centaines de fois plus souvent qu'on
ne publie. Il vaut donc mieux faire le travail au moment de l'écriture, rare, qu'au
moment de la lecture, fréquente.

Et ce n'est que le cas **moyen**. Les pointes de publication multiplient le coût du
fan-out, mais celui-ci peut être **différé** : on met les livraisons en file d'attente et
on accepte que les messages apparaissent un peu plus tard chez les abonnés. Les lectures,
elles, restent rapides puisqu'elles sont servies depuis le cache.

> **À retenir.** La charge s'exprime par quelques **paramètres** (publications/s,
> lectures/s, abonnés moyens) dont les produits donnent le coût de chaque architecture.
> Quand les lectures dominent, déplacer le travail vers l'écriture est payant.

## 3. Le fan-out à l'écriture : des timelines matérialisées (section *Materializing and Updating Timelines*)

**L'idée.** Tenir, pour chaque utilisateur, une structure contenant sa timeline déjà
calculée, comme une **boîte aux lettres**. Quand quelqu'un publie, on cherche ses abonnés
et on dépose le message dans chacune de leurs boîtes. Une lecture n'est plus qu'une
consultation de cache. Le facteur par lequel une requête initiale se multiplie en
requêtes en aval s'appelle le **fan-out** (« éventail »). La timeline est une **donnée
dérivée** : on peut toujours la recalculer depuis les tables, c'est un **cache** du
résultat de la requête, ce qu'on appelle une **vue matérialisée**.

```python
d = Diagram(13.5, 6.0, title="Deux façons de construire une timeline",
            subtitle="en haut : fusion à chaque lecture ; en bas : diffusion (fan-out) à chaque écriture")
d.group(0.2, 3.3, 13.3, 5.9, "à la lecture (requête SQL)", color=INDIGO)
d.box("a1", 1.4, 4.5, "Auteur", kind="user", w=1.1, color=SLATE)
d.box("p1", 4.3, 4.5, "Table posts", sub="1 écriture", color=TEAL, kind="db", w=2.2, h=1.0)
d.box("m1", 7.7, 4.5, "Fusion + tri", sub="1 accès par compte suivi", color=INDIGO, w=2.6)
d.box("l1", 11.6, 4.5, "Lecteur", kind="user", w=1.1, color=SLATE)
d.arrow("a1", "p1", "publie")
d.arrow("p1", "m1", "N suivis", style="dashed")
d.arrow("m1", "l1", "timeline")
d.group(0.2, 0.1, 13.3, 3.1, "à l'écriture (timelines matérialisées)", color=AMBER)
d.box("a2", 1.4, 1.6, "Auteur", kind="user", w=1.1, color=SLATE)
d.box("f2", 4.3, 1.6, "Fan-out", sub="file + travailleurs", color=AMBER, w=2.2)
for i, y in enumerate([2.45, 1.6, 0.75]):
    d.box(f"c{i}", 7.9, y, f"Timeline abonné {i + 1}", color=VIOLET, w=2.6, h=0.6, fontsize=9)
d.box("l2", 11.6, 1.6, "Lecteur", kind="user", w=1.1, color=SLATE)
d.arrow("a2", "f2", "publie")
for i in range(3):
    d.arrow("f2", f"c{i}")
d.arrow("c1", "l2", "1 accès")
d.note(5.95, 2.75, "un message par abonné", fontsize=8.5)
d.show()
```

Implémentons les deux stratégies (classes `rq.PullTimelines` et `rq.PushTimelines`) et
vérifions qu'elles renvoient **exactement les mêmes timelines**, tout en comptant les
opérations élémentaires de chacune. On rejoue les six premiers jours : 40 000 messages et
80 000 lectures de timeline (deux lectures par message, un rapport très prudent par
rapport au cas réel).

```python
pull = rq.PullTimelines(follower, followee, N_USERS, k=50)
push = rq.PushTimelines(follower, followee, N_USERS, k=50)
REPLAY = 40_000
readers = rng.integers(0, N_USERS, 2 * REPLAY)
read_times = np.sort(rng.uniform(0, ts[REPLAY - 1], readers.size))
ok = 0
i_post = 0
for j, (u, t) in enumerate(zip(readers, read_times)):
    while i_post < REPLAY and ts[i_post] <= t:
        for s in (pull, push):
            s.post(int(sender[i_post]), float(ts[i_post]), "")
        i_post += 1
    a, b = pull.timeline(int(u)), push.timeline(int(u))
    if j % 200 == 0:                       # comparaison complète sur un échantillon
        ok += (a == b)
print(f"timelines identiques sur l'échantillon : {ok}/{(readers.size + 199) // 200}")
pd.DataFrame({"écritures": [pull.write_ops, push.write_ops], "lectures": [pull.read_ops, push.read_ops]},
             index=["à la lecture", "à l'écriture"]).assign(total=lambda df: df.sum(axis=1))
```

Les deux structures donnent les mêmes résultats (c'est la définition d'une vue
matérialisée correcte). Mais le travail est réparti très différemment : le fan-out à
l'écriture multiplie les écritures par le nombre moyen d'abonnés (une quarantaine ici)
et ramène chaque lecture à un seul accès. Même avec seulement deux lectures par message,
il fait moins de travail au total.

Le prix à payer : **plus de travail à l'écriture** et un délai avant qu'un message
n'apparaisse chez tous les abonnés. Ce délai devient critique pour certains comptes.

> **À retenir.** Une timeline matérialisée est un cache du résultat de la requête, tenu
> à jour à chaque écriture. On échange du travail à la lecture contre du travail à
> l'écriture, au prix d'une donnée dérivée à maintenir cohérente.

## 4. La loi de puissance des abonnés et les célébrités

Le nombre moyen d'abonnés cache une distribution **extrêmement asymétrique** : la plupart
des comptes ont une poignée d'abonnés, quelques célébrités en ont des dizaines de
millions. Notre générateur suit une loi de Pareto : traçons la fonction de survie
$P(\text{abonnés} \ge x)$ en échelle log-log, où une loi de puissance devient une droite,
sur un graphe plus grand (50 000 comptes).

```python
N_BIG = 50_000
fb, gb = rq.social_graph(N_BIG, mean_follows=50, alpha=1.2, seed=2)
nf_big, ng_big = rq.degree_counts(fb, gb, N_BIG)

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12.5, 4.2))
xs = np.sort(nf_big)
surv = 1.0 - np.arange(xs.size) / xs.size
ax.loglog(xs[xs > 0], surv[xs > 0], color=INDIGO, lw=2, label="abonnés")
ys = np.sort(ng_big)
ax.loglog(ys, 1.0 - np.arange(ys.size) / ys.size, color=TEAL, lw=2, label="abonnements")
ax.axvline(nf_big.mean(), color=GRAY, ls="--", lw=1)
ax.text(nf_big.mean() * 1.1, 0.5, f"moyenne {nf_big.mean():.0f}", color=GRAY, fontsize=9)
ax.set_xlabel("x (échelle log)"); ax.set_ylabel("P(degré ≥ x)"); rq.plain_log_axis(ax)
ax.set_title("Abonnés : une queue en loi de puissance")
ax.legend()

order = np.sort(nf_big)[::-1]
share = np.cumsum(order) / order.sum()
frac = np.arange(1, order.size + 1) / order.size
ax2.plot(frac * 100, share * 100, color=AMBER, lw=2.2)
for p in [0.1, 1, 10]:
    v = share[int(p / 100 * order.size) - 1] * 100
    ax2.plot(p, v, "o", color=ROSE)
    ax2.annotate(f"{p:g} % des comptes : {v:.0f} % du fan-out".replace(".", ","), (p, v), (8, -12), textcoords="offset points", fontsize=9)
ax2.set_xscale("log"); rq.plain_log_axis(ax2, "x"); ax2.set_xlabel("part des comptes les plus suivis (%, log)")
ax2.set_ylabel("part des écritures de fan-out (%)")
ax2.set_title("Les plus gros comptes génèrent l'essentiel du fan-out")
subtitle(ax2, "à taux de publication égal pour tous les comptes")
plt.tight_layout(); plt.show()
print(f"médiane des abonnés : {np.median(nf_big):.0f} ; moyenne : {nf_big.mean():.1f} ; "
      f"maximum : {fmt(nf_big.max())} ({nf_big.max() / N_BIG:.0%} des comptes)")
```

Les abonnements (sarcelle) sont concentrés autour de la moyenne : personne ne suit des
millions de comptes. Les abonnés (indigo) s'étalent sur plusieurs ordres de grandeur, avec
une **médiane bien inférieure à la moyenne** : la moyenne est tirée vers le haut par
quelques comptes géants. Si tout le monde publie au même rythme, les 1 % de comptes les
plus suivis génèrent à eux seuls une grande part des écritures de fan-out.

**Le temps de livraison d'un message de célébrité.** Supposons que l'infrastructure de
fan-out insère 1 million de messages par seconde en tout, et que les publications
d'une célébrité ne puissent pas monopoliser plus de 10 % de cette capacité. Combien de
temps faut-il pour que le dernier abonné reçoive le message ?

```python
capacite = 1_000_000 * 0.10           # insertions par seconde allouées à un seul message
abonnes = np.logspace(1.5, 8.3, 200)
delai = abonnes / capacite
fig, ax = plt.subplots(figsize=(8.5, 4))
ax.loglog(abonnes, delai, color=VIOLET)
for nom, n in [(f"compte moyen ({CHARGE['abonnes_moyen']})", CHARGE["abonnes_moyen"]),
               ("1 million", 1e6), ("100 millions", 1e8)]:
    ax.plot(n, n / capacite, "o", color=ROSE)
    ax.annotate(f"{nom} : {rq.human_number(n / capacite)} s", (n, n / capacite), (8, -14),
                textcoords="offset points", fontsize=9)
ax.axhline(5, color=GRAY, ls="--", lw=1)
ax.text(50, 6.5, "objectif : visible en 5 s", color=GRAY, fontsize=9)
ax.set_xlabel("nombre d'abonnés du compte (log)"); ax.set_ylabel("délai de livraison complète (s, log)")
rq.plain_log_axis(ax)
ax.set_title("Diffuser un message de célébrité prend des minutes")
subtitle(ax, "100 000 insertions/s consacrées à un message")
plt.tight_layout(); plt.show()
```

Le fan-out d'un compte à 100 millions d'abonnés prend un quart d'heure, et mobilise
l'infrastructure pendant ce temps. Pire, ces écritures n'ont rien de « modestes » :
une célébrité qui publie dix fois par jour coûte autant que des millions de comptes
ordinaires. Le cas symétrique existe aussi (un compte qui en suit énormément, très actifs) : sa timeline reçoit tant d'écritures que le livre suggère d'en **ignorer une partie**, puisqu'il ne lira de toute façon qu'un échantillon.

> **À retenir.** Dans une distribution en loi de puissance, la moyenne ne décrit
> personne : le fan-out est dominé par une poignée de comptes extrêmes, qu'il faut
> traiter à part.

## 5. L'approche hybride et le choix d'un seuil

**L'idée.** Diffuser à l'écriture les messages des comptes ordinaires, mais **pas**
ceux des célébrités : ceux-ci restent dans la table des messages et sont fusionnés avec
la timeline matérialisée au moment de la lecture. C'est la solution que décrit le livre.
Un lecteur paie alors un accès au cache plus un accès par célébrité qu'il suit.

**Un petit calcul qui surprend.** Prenons un compte de $F$ abonnés qui publie $p$
messages par seconde, et supposons que chaque abonné lise sa timeline $r$ fois par
seconde. Le diffuser coûte $pF$ écritures par seconde ; ne pas le diffuser coûte $rF$
accès en lecture par seconde (chaque abonné va chercher ses messages à chaque lecture).
Comme on lit bien plus qu'on ne publie ($r > p$), **diffuser est toujours moins cher en
nombre total d'opérations**, quel que soit $F$ ! L'hybride ne se justifie donc pas par
le travail moyen, mais par deux autres critères :

- le **pire cas** : une publication diffusée déclenche $F$ écritures d'un coup, qu'il
  faut absorber en quelques secondes ; les lectures, elles, sont étalées dans le temps ;
- le **coût unitaire** : $F$ écritures dans $F$ timelines dispersées (et répliquées)
  coûtent bien plus cher que relire une petite liste très demandée, qui reste en
  mémoire sur chaque serveur.

Balayons le seuil avec `rq.fanout_costs` sur le grand graphe (une publication par jour
et par compte, 20 lectures de timeline par jour et par compte) et regardons les deux
critères : le travail total et la plus grosse diffusion déclenchée par un seul message.

```python
post_rate, read_rate = 1 / 86400, 20 / 86400
seuils = np.unique(np.logspace(1, np.log10(nf_big.max() + 1), 40).astype(int))
rows = []
for s_ in seuils:
    cst = rq.fanout_costs(fb, gb, N_BIG, post_rate, read_rate, threshold=s_)
    pire = nf_big[nf_big < s_].max()
    rows.append(dict(seuil=s_, pire_fanout=pire, **cst))
H = pd.DataFrame(rows)
base = rq.fanout_costs(fb, gb, N_BIG, post_rate, read_rate)
choix = H.iloc[(H.seuil - 1000).abs().idxmin()]

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4.3))
ax.plot(H.seuil, H.hybride_ecritures, color=AMBER, label="écritures (fan-out)")
ax.plot(H.seuil, H.hybride_lectures, color=INDIGO, label="accès en lecture")
ax.plot(H.seuil, H.hybride_total, color=SLATE, lw=2.8, label="total")
ax.axhline(base["ecriture_total"], color=SLATE, ls=":", lw=1.5, label="fan-out pur (total)")
ax.set_xscale("log"); rq.plain_log_axis(ax, "x")
ax.set_xlabel("seuil d'abonnés au-delà duquel on ne diffuse plus (log)")
ax.set_ylabel("opérations par seconde")
ax.set_title("Travail moyen : le fan-out pur reste le moins cher")
ax.legend(loc="upper right")

ax2.plot(H.pire_fanout, H.hybride_total, "o-", color=VIOLET, ms=4, lw=1.5)
for s_lab in [10, 100, 1000, 10000]:
    r_ = H.iloc[(H.seuil - s_lab).abs().idxmin()]
    ax2.annotate(f"seuil {r_.seuil:.0f}", (r_.pire_fanout, r_.hybride_total), (8, 6),
                 textcoords="offset points", fontsize=9, color=VIOLET)
ax2.plot(nf_big.max(), base["ecriture_total"], "s", color=AMBER, ms=9, label="fan-out pur")
ax2.set_xscale("log"); rq.plain_log_axis(ax2, "x")
ax2.set_xlabel("plus grosse diffusion déclenchée par un message (log)")
ax2.set_ylabel("travail total (opérations par seconde)")
ax2.set_title("Le seuil échange du travail moyen contre un pire cas borné")
ax2.legend(loc="upper right")
plt.tight_layout(); plt.show()
pd.DataFrame({"lecture pure": [base["lecture_total"], 1],
              "fan-out pur": [base["ecriture_total"], nf_big.max()],
              f"hybride (seuil {choix.seuil:.0f})": [choix.hybride_total, choix.pire_fanout]},
             index=["opérations/s (moyenne)", "plus grosse diffusion d'un message"]).round(1)
```

Le graphique de gauche confirme le calcul : en nombre d'opérations, abaisser le seuil
ne fait qu'augmenter le travail total (les accès en lecture grimpent plus vite que les
écritures ne baissent). La courbe d'écritures a une forme en marches d'escalier : chaque
célébrité qui repasse dans le régime « diffusé » ajoute d'un coup tout son fan-out.

Le graphique de droite montre ce qu'on achète avec ce surcoût : chaque point est un
seuil, placé selon la **plus grosse diffusion** qu'un seul message peut déclencher et le
travail total. Avec un seuil d'un millier d'abonnés, aucune publication ne provoque plus
d'un millier d'écritures (contre des dizaines de milliers pour le plus gros compte de
notre petit graphe, et des dizaines de millions dans un vrai réseau social). Le prix est
un nombre d'accès en lecture plusieurs fois plus grand ; mais ces accès portent sur
quelques listes très demandées, qui restent en mémoire, alors que le fan-out écrit dans
des timelines dispersées. Si l'on donne à une écriture dispersée un coût unitaire plus
élevé qu'à la relecture d'une liste chaude, l'hybride devient gagnant même en moyenne
(exercice 2.2).

Vérifions enfin l'implémentation hybride (`rq.HybridTimelines`) sur le petit graphe :
elle doit renvoyer exactement le résultat de la requête SQL.

```python
seuil = 200
hyb = rq.HybridTimelines(follower, followee, N_USERS, k=50, threshold=seuil)
for i in range(N_POSTS):
    hyb.post(int(sender[i]), float(ts[i]), "")
users_test = rng.choice(N_USERS, 200, replace=False)
sql_ok = sum([r[0] for r in con.execute(TIMELINE_SQL, (int(u),)).fetchall()]
             == [int(np.searchsorted(ts, m[0])) for m in hyb.timeline(int(u))] for u in users_test)
print(f"hybride = requête SQL sur {sql_ok}/200 timelines")
print(f"{sum(hyb.celebrity)} célébrités (≥ {seuil} abonnés) ; pire fan-out d'un message : "
      f"{max(n for n in n_followers if n < seuil)} écritures au lieu de {n_followers.max()}")
```

> **À retenir.** L'approche hybride diffuse les comptes ordinaires et fusionne les
> célébrités à la lecture. En nombre brut d'opérations elle ne gagne pas (on lit plus
> qu'on n'écrit), mais elle **borne le pire cas** d'une publication, ce qui compte pour
> la latence de livraison et le dimensionnement.

## 6. Décrire la charge (section *Understanding Load*)

Toute discussion sur la croissance (« que se passe-t-il si la charge double ? ») suppose
d'abord de **décrire la charge actuelle** avec quelques **paramètres de charge**
pertinents. Lesquels ? Cela dépend de l'architecture : requêtes par seconde, rapport
lectures/écritures, nombre d'utilisateurs simultanés, taux de réussite d'un cache,
nombre d'éléments par utilisateur. Dans l'étude de cas, la **distribution des abonnés**
pondérée par la fréquence de publication est le paramètre décisif, parce qu'elle fixe la
charge de fan-out.

Autre point : la charge **varie dans le temps**. Une moyenne journalière masque un cycle
quotidien et des pics d'événements. Simulons une semaine de publications (cycle jour/nuit
et deux événements ponctuels) et comparons la moyenne, le 99e percentile minute par minute
et le maximum.

```python
minutes = np.arange(7 * 24 * 60)
heure = (minutes / 60) % 24
moyenne_cible = CHARGE["publications_par_s"]
profil = 1 + 0.6 * np.sin(2 * np.pi * (heure - 14) / 24 + np.pi / 2)    # pic vers 14 h
evenements = np.zeros_like(profil)
for jour, h, amp in [(2, 21, 6.0), (5, 19, 3.0)]:                      # finale, élection...
    t0 = (jour * 24 + h) * 60
    evenements += amp * np.exp(-0.5 * ((minutes - t0) / 12) ** 2)
lam = moyenne_cible * (profil + evenements) / (profil + evenements).mean()
debit = rng.poisson(lam * 60) / 60

fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(minutes / 1440, debit / 1e3, color=INDIGO, lw=0.8)
for val, nom, col in [(debit.mean(), "moyenne", TEAL), (np.percentile(debit, 99), "99e percentile des minutes", AMBER),
                      (debit.max(), "maximum", ROSE)]:
    ax.axhline(val / 1e3, color=col, ls="--", lw=1.4)
    ax.text(0.9, val / 1e3 + 0.4, f"{nom} : {val / 1e3:.1f} k/s".replace(".", ","), color=col, va="bottom",
            fontsize=9, bbox=dict(fc="white", ec="none", alpha=0.85, pad=1))
ax.set_xlim(0, 7); ax.set_ylim(0, debit.max() / 1e3 * 1.12); ax.set_xlabel("jour"); ax.set_ylabel("publications par seconde (milliers)")
ax.set_title("La capacité se dimensionne sur les pics, pas sur la moyenne")
subtitle(ax, "semaine simulée : cycle quotidien et deux événements exceptionnels")
plt.tight_layout(); plt.show()
print(f"rapport pic / moyenne : {debit.max() / debit.mean():.1f}")
```

Le rapport entre pic et moyenne atteint ici un facteur de six à sept (le livre cite 150 000 messages/s en pointe pour 5 800 en moyenne, un facteur 26).
Selon le système, on dimensionne pour le pic (si on ne peut pas différer le travail) ou
pour un niveau intermédiaire en **absorbant** le pic dans une file d'attente (le fan-out
peut prendre du retard, la lecture non). Une fois la charge décrite, on peut se poser
les deux questions du livre : à ressources constantes, comment la performance se
dégrade-t-elle quand la charge augmente ? et combien de ressources ajouter pour garder
la même performance ? Il faut pour cela savoir **mesurer la performance** : c'est l'objet
du cours 2.

> **À retenir.** Décrire la charge, c'est choisir les quelques paramètres qui
> déterminent le travail de l'architecture (débits, rapports, distributions) et regarder
> leurs **pics** et leurs **cas extrêmes**, pas seulement leur moyenne.

## Synthèse

- La timeline d'un réseau social se calcule **à la lecture** (jointure et tri, coût
  proportionnel au nombre de comptes suivis) ou **à l'écriture** (fan-out vers des
  timelines matérialisées, coût proportionnel au nombre d'abonnés).
- Le **rapport lectures/écritures** décide : quand on lit bien plus qu'on n'écrit,
  pré-calculer est payant. Une timeline matérialisée est une donnée dérivée, un cache
  qu'il faut tenir à jour.
- Les abonnés suivent une **loi de puissance** : la moyenne masque des célébrités dont le
  fan-out coûte des minutes. L'**approche hybride** les traite à part : elle ne réduit
  pas le travail moyen, mais elle borne le pire cas.
- Décrire la charge, c'est choisir des **paramètres de charge** et regarder leurs pics et
  leurs extrêmes.
