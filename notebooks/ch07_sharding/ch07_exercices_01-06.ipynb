## Exercice 7.1 : choisir une clé de partitionnement

**Énoncé.** Pour chacune des trois applications suivantes, on dispose de 8
shards. Proposer une clé de partitionnement et un schéma (par plage
de clés ou par hachage), en justifiant par les requêtes à servir.

1. **Capteurs.** 500 capteurs envoient chacun une mesure toutes les 10 minutes. Requêtes :
   (a) « toutes les mesures de tous les capteurs entre 10 h et 11 h hier » ; (b)
   « l'historique complet du capteur 42 ». On compare trois clés : l'horodatage seul (par
   plage, un shard par jour), le couple (capteur, horodatage) par plage, et le
   couple (capteur, horodatage) avec hachage du seul identifiant de capteur.
2. **Commandes d'un site marchand.** Requêtes : afficher une commande ; lister les
   commandes d'un client. Clé : identifiant de commande haché, ou identifiant de client
   haché ?
3. **Messages d'un réseau social**, clé = identifiant de l'auteur hachée. La popularité
   des comptes suit une loi de Zipf. La répartition des **lectures** est-elle équitable ?

Vérifier chaque réponse en mesurant le déséquilibre (charge maximale / charge moyenne)
et le nombre de shards qu'une requête doit interroger.

**Rappels.** Par plage de clés, les clés voisines sont ensemble : les requêtes de plage
sont efficaces mais des écritures sur des clés voisines (horodatages) créent un point
chaud. Par hachage, la charge s'étale mais l'ordre est perdu ; seule la partie hachée de
la clé décide de le shard, le reste peut servir à trier (cours 1, sections
*Sharding by Key Range* et
*Sharding by Hash of Key*). Hacher ne règle pas
le cas d'une clé très demandée (cours 3, section *Skewed Workloads and Relieving Hot
Spots*).

**Solution.**

*1. Capteurs.* Avec l'horodatage seul, toutes les écritures d'une journée vont dans
le même shard : déséquilibre en écriture égal à 8 (un shard travaille, sept
dorment). La requête (a) est idéale (un seul shard, un parcours de plage) ; la
requête (b) touche les 8 shards. Avec (capteur, horodatage), les écritures
s'étalent, (b) devient un parcours de plage dans un seul shard, mais (a) exige
**une requête de plage par capteur**, soit 500 requêtes réparties sur tous les
shards. Comme on écrit en continu et qu'on lit l'historique d'un capteur, c'est
le bon choix ; la requête (a), plus rare, se paie. Par plage ou par hachage de
l'identifiant de capteur, les deux variantes se valent ici ; la plage donne un meilleur
équilibre parce que les identifiants sont denses et de charge égale, le hachage de 500
valeurs seulement laisse un déséquilibre visible (15 %, voir l'exercice 7.2).

*2. Commandes.* Hacher l'identifiant de **client** place toutes les commandes d'un client
ensemble : « lister les commandes d'un client » touche un shard, et l'affichage
d'une commande aussi pourvu que l'identifiant de commande contienne celui du client (clé
composée (client, commande), le client servant de clé de partitionnement et la commande
de clé de tri). Hacher l'identifiant de commande équilibre parfaitement mais oblige à
interroger tous les shards (ou un index secondaire global) pour la liste d'un
client.

*3. Réseau social.* Le hachage répartit les **comptes** équitablement, pas la
**charge** : les lectures d'un compte célèbre arrivent toutes sur
le même shard. Si le compte le plus lu reçoit une part des lectures supérieure à 1/8, aucun
schéma de partitionnement par auteur ne peut équilibrer ; il faut des répliques, un cache,
ou un traitement spécial des clés chaudes.

**Vérification.** Huit jours de mesures simulées (`P.sensor_readings`), puis 100 000
comptes avec des lectures en loi de Zipf d'exposant 1,1.

```python
N = 8
sid, t = P.sensor_readings(500, 8 * 86400, 1 / 600, seed=1)
jour = (t // 86400).astype(int)
schemas = {
    "horodatage (plage, 1 par jour)": jour,
    "(capteur, horodatage) par plage": (sid * N) // 500,
    "(capteur, horodatage), capteur haché": (P.mix64(sid) % np.uint64(N)).astype(int),
}
rows = []
for nom, sh in schemas.items():
    imb = np.mean([P.imbalance(P.shard_loads(sh[jour == d], N)) for d in range(8)])
    req_a = len(np.unique(sh[(jour == 6) & ((t % 86400) // 3600 == 10)]))
    req_b = len(np.unique(sh[sid == 42]))
    rows.append({"clé": nom, "déséquilibre en écriture (par jour)": imb,
                 "shards lus, requête (a)": req_a,
                 "requêtes de plage, (a)": 1 if nom.startswith("horo") else 500,
                 "shards lus, requête (b)": req_b})
tab = pd.DataFrame(rows).set_index("clé")
print(f"{fmt(len(sid))} mesures simulées")
tab
```

```python
cmd_client = rng.integers(0, 20_000, 200_000)          # client de chaque commande
cmd_id = np.arange(cmd_client.size)
par_cmd = (P.mix64(cmd_id) % np.uint64(N)).astype(int)
par_cli = (P.mix64(cmd_client) % np.uint64(N)).astype(int)
clients = rng.choice(20_000, 500, replace=False)
lues_cmd = np.mean([len(np.unique(par_cmd[cmd_client == c])) for c in clients])
lues_cli = np.mean([len(np.unique(par_cli[cmd_client == c])) for c in clients])
print(f"commandes par client : {cmd_client.size / 20_000:.0f} en moyenne")
print(f"lister les commandes d'un client : {lues_cmd:.2f} shards lus (clé = commande) "
      f"contre {lues_cli:.2f} (clé = client)")
print(f"déséquilibre en volume : {P.imbalance(P.shard_loads(par_cmd, N)):.3f} (commande), "
      f"{P.imbalance(P.shard_loads(par_cli, N)):.3f} (client)")
```

```python
n_comptes = 100_000
lectures = P.zipf_requests(n_comptes, 1_000_000, s=1.1, seed=2)
part_compte = (P.mix64(np.arange(n_comptes)) % np.uint64(N)).astype(int)
charge = P.shard_loads(part_compte[lectures], N)
comptes = P.shard_loads(part_compte, N)
part_top = np.mean(lectures == 0)

fig, axes = plt.subplots(1, 2, figsize=(11, 3.8))
for ax, val, titre, c in [(axes[0], comptes, "comptes par shard", TEAL),
                          (axes[1], charge, "lectures par shard", ROSE)]:
    ax.bar(range(N), val / val.mean(), color=c)
    ax.axhline(1, color=GRAY, ls=":", lw=1)
    ax.set_xlabel("shard"); ax.set_ylabel("charge / moyenne")
    ax.set_title(titre + f" : max/moy = {P.imbalance(val):.2f}".replace(".", ","))
axes[1].bar(part_compte[0], charge[part_compte[0]] / charge.mean(), color=AMBER, label="contient le compte n° 1")
axes[1].legend(loc="upper left")
finish(fig, "Hacher l'auteur équilibre les comptes, pas les lectures")
plt.show()
print(f"part des lectures du compte le plus populaire : {part_top:.1%} (1/8 = 12,5 %)")
```

Les chiffres confirment le raisonnement. Pour les capteurs, l'horodatage seul donne un
déséquilibre de 8 (toute la journée sur un shard) ; les deux clés composées
l'éliminent, la plage sur l'identifiant de capteur étant presque parfaite et le hachage
de 500 identifiants laissant un écart de quelques pour cent à une quinzaine de pour cent,
que l'exercice suivant explique. La requête (a) touche alors tous les
shards et coûte 500 parcours de plage, la requête (b) un seul. Pour les commandes, la clé
« client » ramène la liste d'un client à un seul shard contre 5,75 en moyenne avec la
clé « commande » (10 commandes par client, réparties au hasard sur 8). Pour le réseau social, le compte le plus lu attire à lui seul 13 %
des lectures : son shard porte largement plus que la moyenne, quel que soit le
hachage.

> **À retenir.** On choisit la clé de partitionnement d'après les **requêtes** : la
> requête fréquente doit toucher un seul shard, les écritures doivent s'étaler.
> Une clé composée (clé de partitionnement, clé de tri) concilie souvent les deux. Et
> aucun schéma ne répartit une charge concentrée sur une seule clé.

## Exercice 7.2 : le hachage n'équilibre qu'en moyenne

**Énoncé.** On hache $m$ clés de charge égale dans $n$ shards : chaque
shard reçoit en moyenne $\lambda = m/n$ clés.

1. Quelle loi suit le nombre de clés d'un shard ? Donner son écart-type relatif.
2. Le maximum de $n$ telles variables vaut environ $\lambda + z_n \sqrt{\lambda}$, avec
   $z_n \approx \sqrt{2 \ln n}$. En déduire le nombre de clés par shard nécessaire
   pour que le déséquilibre max/moyenne reste sous 1,1, pour $n = 10$, 100 et 1 000.
3. Vérifier par simulation. Que dire d'un système qui ne place que quelques dizaines
   de clés (ou de capteurs, ou de clients) par shard ?

**Rappels.** Une bonne fonction de hachage rend les positions des clés indépendantes et
uniformes (cours 1, section *Sharding by Hash of Key*) : le
placement des clés est alors exactement l'expérience des « balles jetées au hasard dans
des urnes ».

**Solution.**

*1.* Chaque clé tombe dans un shard donné avec probabilité $1/n$, indépendamment :
le nombre de clés suit une loi binomiale $\mathcal{B}(m, 1/n)$, quasi poissonnienne de
paramètre $\lambda$. Son écart-type vaut $\sqrt{\lambda}$, soit un écart **relatif** de
$1/\sqrt{\lambda}$ : 100 clés par shard donnent déjà ±10 % d'un shard à l'autre.

*2.* On veut $z_n\sqrt{\lambda} \le 0{,}1\,\lambda$, soit $\lambda \ge (z_n / 0{,}1)^2 =
200 \ln n$ : environ **460** clés par shard pour $n = 10$, **920** pour $n = 100$,
**1 380** pour $n = 1\,000$. L'approximation $\sqrt{2 \ln n}$ surestime un peu le
maximum pour les petits $n$ ; une version plus fine retranche
$(\ln \ln n + \ln 4\pi) / (2\sqrt{2\ln n})$, ce qui abaisse les seuils (environ 190, 560
et 970).

*3.* Avec quelques dizaines de clés par shard, l'écart dépasse 50 % :
le shard le plus chargé porte une fois et demie la moyenne. C'est le cas des 500
capteurs de l'exercice 7.1 (62 par shard). Le hachage n'est un bon égaliseur que
s'il y a **beaucoup** de clés de poids comparables.

```python
def z_simple(n):
    return math.sqrt(2 * math.log(n))

def z_fin(n):
    z = z_simple(n)
    return z - (math.log(math.log(n)) + math.log(4 * math.pi)) / (2 * z)

lams = np.unique(np.round(np.logspace(1, 3.7, 14)).astype(int))
fig, ax = plt.subplots(figsize=(9, 4.4))
rows = []
for n, c in [(10, TEAL), (100, INDIGO), (1000, ROSE)]:
    sim = [np.median([(lambda x: x.max() / x.mean())(np.bincount(rng.integers(0, n, lam * n), minlength=n))
                      for _ in range(30)]) for lam in lams]
    ax.plot(lams, sim, "o", color=c, ms=5, label=f"simulation, n = {n}")
    ax.plot(lams, 1 + z_fin(n) / np.sqrt(lams), color=c, lw=1.5, alpha=0.8)
    rows.append({"n": n, "seuil 200 ln n": 200 * math.log(n), "seuil affiné": (z_fin(n) / 0.1) ** 2,
                 "déséquilibre simulé à λ = 1000": np.median([(lambda x: x.max() / x.mean())(
                     np.bincount(rng.integers(0, n, 1000 * n), minlength=n)) for _ in range(30)])})
ax.axhline(1.1, color=AMBER, ls="--", lw=1.2)
ax.text(11, 1.12, "objectif : max/moyenne < 1,1", color=AMBER, fontsize=9)
ax.set_xscale("log")
ax.set_xlabel("clés par shard (λ)")
ax.set_ylabel("déséquilibre max / moyenne")
ax.set_title("Il faut des centaines de clés par shard pour un équilibre à 10 %")
subtitle(ax, "points : médiane de 30 tirages ; traits : 1 + z_n / √λ avec la correction fine de z_n")
ax.legend()
plt.tight_layout(); plt.show()
pd.DataFrame(rows).set_index("n").round(3)
```

La formule affinée suit les points simulés de près dès une centaine de clés par
shard (en dessous, la loi de Poisson est asymétrique et le maximum dépasse un peu la
formule), et les seuils se lisent sur la figure :
à λ = 1 000, le déséquilibre reste sous 1,1 pour n = 10 et n = 100, et l'atteint tout juste
pour n = 1 000 (1,10), conformément au seuil affiné d'environ 970 clés. Le déséquilibre décroît comme $1/\sqrt{\lambda}$ : pour le diviser par
deux, il faut quatre fois plus de clés par shard.

> **À retenir.** Hacher $m$ clés dans $n$ shards laisse un écart relatif de l'ordre
> de $\sqrt{2\ln n / \lambda}$ entre le shard le plus chargé et la moyenne. Avec peu de
> clés par shard, ou des clés de poids très inégaux, le hasard seul crée des points
> chauds.

## Exercice 7.3 : combien de clés bougent, et combien de jetons par nœud ?

**Énoncé.**

1. Avec l'affectation « haché mod N », montrer que passer de $N$ à $N+1$ nœuds déplace
   une fraction $N/(N+1)$ des clés. Quelle est la fraction minimale à déplacer pour
   rétablir l'équilibre ?
2. Avec un anneau de hachage où chacun des $n$ nœuds tire $v$ jetons au hasard, la part
   de l'anneau possédée par un nœud est la somme de $v$ arcs. Montrer que son écart-type
   relatif vaut environ $1/\sqrt{v}$. Combien de jetons faut-il pour que la part d'un nœud
   s'écarte typiquement de moins de 10 % de $1/n$ ?
3. Vérifier ces deux résultats avec `P.mod_n_assign`, `P.FixedShardCluster` et
   `P.HashRing`.

**Rappels.** Cours 2 (section *Sharding by Hash of Key*) :
l'affectation modulo N est la mauvaise façon de faire ; un nombre fixe de
shards ne déplace que des shards entiers ; l'anneau à jetons aléatoires
(hachage cohérent à la Cassandra)
découpe des plages existantes pour le nouveau nœud.

**Solution.**

*1.* Prenons le haché $h$ uniforme sur $[0, N(N+1))$ (un multiple commun de $N$ et
$N+1$ ; le cas général s'en déduit à une erreur négligeable près). La clé reste en place
si $h \bmod N = h \bmod (N+1) = r$. Par le théorème des restes chinois ($N$ et $N+1$ sont
premiers entre eux), cela équivaut à $h \equiv r \pmod{N(N+1)}$, donc $h = r$ avec
$0 \le r < N$ : seules $N$ valeurs sur $N(N+1)$ restent, une fraction $1/(N+1)$. La
fraction déplacée vaut donc $N/(N+1)$ : 75 % de 3 à 4 nœuds, 91 % de 10 à 11. Or le
minimum est l'inverse : le nouveau nœud doit recevoir sa part, $1/(N+1)$, et rien
d'autre n'a besoin de bouger.

*2.* Les $nv$ jetons uniformes découpent l'anneau en arcs de longueur quasi
exponentielle de moyenne $1/(nv)$ et de même écart-type. La part d'un nœud est la somme
de ses $v$ arcs (presque indépendants) : moyenne $v/(nv) = 1/n$, écart-type
$\sqrt{v}/(nv)$, soit un écart **relatif** de $1/\sqrt{v}$. Pour descendre sous 10 %, il
faut $v \ge 100$ jetons par nœud ; avec un seul jeton, l'écart typique est de 100 % : un
nœud peut posséder trois ou quatre fois sa part. Cassandra a longtemps utilisé 256 jetons par nœud par défaut (écart typique de 6 %), et en utilise 16 depuis sa version 4.0, compromis entre équilibre et coût de gestion.

```python
keys = np.arange(200_000)
rows = []
for n in [3, 5, 10, 20]:
    modn = P.moved_fraction(P.mod_n_assign(keys, n), P.mod_n_assign(keys, n + 1))
    fixe = P.FixedShardCluster(840, [f"n{i}" for i in range(n)])
    av = fixe.owner_names(keys); fixe.add_node("nouveau"); fx = P.moved_fraction(av, fixe.owner_names(keys))
    ring = P.HashRing([f"n{i}" for i in range(n)], vnodes=128, seed=n)
    av = ring.node_of(keys); ring.add_node("nouveau"); rg = P.moved_fraction(av, ring.node_of(keys))
    rows.append({"N -> N+1": f"{n} -> {n + 1}", "théorie mod N : N/(N+1)": n / (n + 1), "mod N mesuré": modn,
                 "840 shards fixes": fx, "anneau, 128 jetons": rg, "optimum 1/(N+1)": 1 / (n + 1)})
tab = pd.DataFrame(rows).set_index("N -> N+1")
tab
```

```python
vs = [1, 2, 4, 8, 16, 32, 64, 128, 256]
ecarts = []
for v in vs:
    rel = []
    for seed in range(30):
        r = P.HashRing([f"n{i}" for i in range(10)], vnodes=v, seed=seed)
        rel += list(np.array(list(r.ownership().values())) * 10 - 1)
    ecarts.append((np.std(rel), np.max(np.abs(rel))))
ecarts = np.array(ecarts)
fig, ax = plt.subplots(figsize=(9, 4.2))
ax.loglog(vs, ecarts[:, 0], "o-", color=INDIGO, label="écart-type relatif mesuré")
ax.loglog(vs, ecarts[:, 1], "s--", color=ROSE, ms=5, label="pire écart observé (300 nœuds)")
ax.loglog(vs, 1 / np.sqrt(vs), color=GRAY, lw=1.2, label="théorie : 1/√v")
ax.axhline(0.1, color=AMBER, ls=":", lw=1.2)
ax.text(1.1, 0.108, "10 %", color=AMBER, fontsize=9)
ax.set_xlabel("jetons par nœud (v)"); ax.set_ylabel("écart relatif à la part équitable")
ax.set_title("La part d'un nœud sur l'anneau se resserre en 1/√v")
subtitle(ax, "10 nœuds, 30 anneaux tirés au hasard pour chaque v")
ax.legend()
plt.tight_layout(); plt.show()
```

Le tableau retrouve $N/(N+1)$ pour le modulo, alors que le nombre fixe de
shards et l'anneau déplacent à peu près l'optimum $1/(N+1)$ (le nombre fixe exactement à
l'arrondi près, puisque 840 est divisible par 3, 4, 5, 6, 10, 20 et 21 ; l'anneau à
quelques points près, parce que la part du nouveau nœud est elle-même aléatoire). La
figure confirme la loi en $1/\sqrt{v}$ : l'écart-type mesuré passe sous 10 % vers
$v = 100$. Le pire nœud, lui, s'écarte environ trois fois plus que l'écart-type : avec
16 jetons, le pire des 300 nœuds tirés porte encore environ deux fois sa part.

> **À retenir.** Le modulo déplace presque tout ($N/(N+1)$), les schémas conçus pour le
> rééquilibrage déplacent le minimum ($1/(N+1)$). Sur un anneau à jetons aléatoires,
> l'équilibre s'achète par le nombre de jetons : l'écart relatif vaut $1/\sqrt{v}$.

## Exercice 7.4 : combien de shards fixes pour un jeu de données qui croît ?

**Énoncé.** Une base démarre avec 2 To et devrait atteindre 200 To en cinq ans
(croissance exponentielle régulière). L'équipe d'exploitation juge qu'un shard
se gère bien entre 10 et 50 Go : en dessous, le surcoût fixe par
shard (fichiers, métadonnées, mémoire, coordination) domine ; au-dessus, déplacer
ou reconstruire un shard devient long. Le nombre de shards est fixé à la
création.

1. Quel nombre de shards respecte la borne haute dans cinq ans ? Quelle taille
   ont-ils au départ ? Peut-on rester dans la fourchette pendant toute la période ?
2. Le cluster passera de 6 à environ 120 nœuds. Pourquoi préférer 5 040 à 4 096 ?
3. Alternative : démarrer avec un nombre de shards adapté à 2 To et repartitionner
   (scinder chaque shard en deux) quand la borne haute est atteinte. Combien de
   scissions générales en cinq ans, et quel volume réécrit au total ?

**Rappels.** Avec un nombre fixe de shards, la taille de chacun est proportionnelle
au volume total ; un mauvais choix initial impose un repartitionnement coûteux (cours 2,
section *Fixed number of shards*).

**Solution.**

*1.* Borne haute à 200 To : $200\,000 / 50 = 4\,000$ shards au moins. Au départ,
$2\,000 / 4\,000 = 0{,}5$ Go chacun, vingt fois sous la borne basse. La fourchette
10-50 Go ne couvre qu'un facteur 5 alors que les données croissent d'un facteur 100 :
**aucun nombre fixe ne convient sur toute la période**. Avec une croissance de
$100^{1/5} \approx 2{,}5$ par an, la taille entre dans la fourchette au bout de
$\ln(20)/\ln(2{,}51) \approx 3{,}3$ ans : on accepte trois ans de shards trop petits
(le surcoût fixe est le moindre mal).

*2.* On veut répartir les shards équitablement quel que soit le nombre de nœuds :
5 040 = 2⁴·3²·5·7 a 60 diviseurs (6, 7, 8, 9, 10, 12, 14, 15, 16, 18, 20, 21, 24, 28,
30... 120), alors que 4 096 n'a que 13 diviseurs, tous des puissances de deux : avec 6,
10 ou 120 nœuds, certains nœuds portent un shard de plus que d'autres, là où 5 040
se répartit exactement (100 ne divise pas 5 040 non plus : 50 ou 51 par nœud). Avec 4 096
shards sur 100 nœuds, c'est 41 contre 40, soit 2,5 % d'écart ; sur 120 nœuds, 35
contre 34 (3 %). L'écart est modeste avec autant de shards : il devient sensible
quand il y en a peu par nœud.

*3.* 2 To / 50 Go = 40 shards au départ (pleins) ; chaque scission générale double
leur nombre. Il en faut $\lceil \log_2 100 \rceil = 7$ pour suivre (40 × 2⁷ = 5 120). Une
scission réécrit tout le volume présent à ce moment-là : environ 4, 8, 16... 128 To,
soit à peu près **250 To réécrits** en cinq ans, plus que la taille finale. C'est le prix
de shards toujours bien dimensionnés ; les systèmes à partitionnement dynamique
étalent ce coût en scindant chaque shard au fil de l'eau plutôt que tous à la
fois.

```python
ans = np.linspace(0, 5, 300)
volume = 2_000 * 100 ** (ans / 5)          # Go

fig, ax = plt.subplots(figsize=(9.5, 4.4))
ax.axhspan(10, 50, color=GREEN, alpha=0.12)
ax.text(1.75, 12, "zone confortable 10-50 Go", color=GREEN, fontsize=9)
for n, c in [(64, ROSE), (512, AMBER), (5040, INDIGO)]:
    ax.plot(ans, volume / n, color=c, label=f"{n} shards fixes")
# repartitionnement : on double quand on dépasse 50 Go
n_dyn, taille, scissions, reecrit = 40, [], [], 0.0
for a, v in zip(ans, volume):
    if v / n_dyn > 50:
        reecrit += v
        n_dyn *= 2
        scissions.append(a)
    taille.append(v / n_dyn)
ax.plot(ans, taille, color=TEAL, lw=1.8, label="40 au départ, doublés à 50 Go")
for a in scissions:
    ax.axvline(a, color=TEAL, ls=":", lw=0.8)
ax.set_yscale("log")
ax.set_xlabel("années"); ax.set_ylabel("taille par shard (Go)")
ax.set_title("Un nombre fixe ne suit pas une croissance d'un facteur 100")
subtitle(ax, "volume : 2 To -> 200 To en 5 ans ; pointillés : scissions générales")
ax.legend(loc="upper left", fontsize=8.5)
plt.tight_layout(); plt.show()

entree = math.log(20) / math.log(100 ** 0.2)
print(f"5 040 : taille initiale {2000 / 5040:.2f} Go, entrée dans la zone après {entree:.1f} ans")
print(f"scissions générales : {len(scissions)}, volume réécrit : {reecrit / 1000:.0f} To")
div = lambda n: sum(1 for d in range(1, n + 1) if n % d == 0)
print(f"diviseurs : 5040 -> {div(5040)}, 4096 -> {div(4096)}, 840 -> {div(840)}")
for nodes in [6, 10, 100, 120]:
    print(f"{nodes:>3} nœuds : 4096 -> {4096 // nodes}..{-(-4096 // nodes)} par nœud ; "
          f"5040 -> {5040 // nodes}..{-(-5040 // nodes)}")
```

La simulation retrouve les sept scissions générales et un volume réécrit cumulé du même
ordre que l'estimation (un peu au-dessus de 200 To, car chaque scission a lieu dès que
le seuil est franchi). La figure montre le dilemme : la courbe des 5 040
shards n'entre dans la zone confortable qu'au bout de trois ans et quelques, et
une valeur plus faible (512, 64) en sort par le haut bien avant la fin.

> **À retenir.** Un nombre fixe de shards suppose de connaître l'ordre de grandeur
> final des données. On le choisit pour la taille cible (borne haute), en acceptant des
> shards trop petits au début, et riche en diviseurs pour s'adapter à tout nombre
> de nœuds. Si la croissance est imprévisible, le partitionnement dynamique est plus
> adapté.

## Exercice 7.5 : fusionner les petits shards, sans osciller

**Énoncé.** La classe `P.DynamicRangeSharding` scinde un shard qui dépasse
`max_keys` clés, mais ne fusionne jamais.

1. Écrire une sous-classe qui supporte la suppression d'une clé et **fusionne**
   un shard tombé sous `min_keys` avec son plus petit voisin (si le résultat ne
   dépasse pas `max_keys`).
2. Insérer 40 000 clés (`max_keys = 1 000`), puis purger 90 % des clés anciennes (les
   75 % inférieures de l'espace de clés, comme une purge de données périmées).
   Comparer le nombre de shards et leur taille sans fusion et avec fusion.
3. Montrer qu'avec `min_keys` supérieur à `max_keys / 2`, une suite alternée
   d'insertions et de suppressions autour d'une frontière provoque une **oscillation**
   scission / fusion, et mesurer le volume réécrit. Quelle règle en déduire ?

**Rappels.** Le partitionnement dynamique scinde un shard trop gros et fusionne
des shards voisins devenus trop petits, comme les nœuds d'un B-tree ; chaque
opération réécrit les données concernées (cours 2, section
*Rebalancing key-range sharded data*).

**Solution.** *1.* Implémentation ci-dessous : la suppression retire la position de la
liste triée de son shard, puis teste la fusion ; une fusion supprime la frontière
entre les deux shards et réécrit leurs clés.

```python
import bisect


class MergingSharding(P.DynamicRangeSharding):
    """Partitionnement dynamique avec scissions (héritées) et fusions."""

    def __init__(self, max_keys, min_keys, **kw):
        super().__init__(max_keys, **kw)
        self.min_keys = min_keys
        self.merges = 0
        self.n_splits = 0

    def insert(self, key):
        n0 = self.n_shards
        i = super().insert(key)
        self.n_splits += self.n_shards - n0
        return i

    def delete(self, key):
        pos = self.position(key)
        i = bisect.bisect_right(self.bounds, pos)
        part = self.parts[i]
        j = bisect.bisect_left(part, pos)
        if j == len(part) or part[j] != pos:
            return
        part.pop(j)
        self._maybe_merge(i)

    def _maybe_merge(self, i):
        if self.n_shards == 1 or len(self.parts[i]) >= self.min_keys:
            return
        voisins = [k for k in (i - 1, i + 1) if 0 <= k < self.n_shards]
        k = min(voisins, key=lambda k: len(self.parts[k]))
        a, b = min(i, k), max(i, k)
        fusion = self.parts[a] + self.parts[b]
        if len(fusion) > self.max_keys:
            return
        self.parts[a:b + 1] = [fusion]
        self.writes[a:b + 1] = [self.writes[a] + self.writes[b]]
        del self.bounds[a]
        self.rewritten += len(fusion)
        self.merges += 1


cles = rng.permutation(200_000)[:40_000]
anciennes = [int(k) for k in cles if k < 150_000]
purge = anciennes[: int(0.9 * len(anciennes))]
etats = {}
for nom, mn in [("sans fusion", 0), ("fusion sous 250 clés", 250)]:
    d = MergingSharding(1_000, mn)
    for k in cles:
        d.insert(int(k))
    avant = d.n_shards
    for k in purge:
        d.delete(k)
    etats[nom] = d
    print(f"{nom:>22} : {avant} shards -> {d.n_shards} après la purge, "
          f"médiane {np.median(d.sizes()):.0f} clés, fusions : {d.merges}")

fig, axes = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for ax, (nom, d), c in zip(axes, etats.items(), [ROSE, TEAL]):
    ax.bar(range(d.n_shards), d.sizes(), color=c, width=0.85)
    ax.axhline(1_000, color=GRAY, ls=":", lw=1)
    ax.set_title(f"{nom} : {d.n_shards} " + "shards")
    ax.set_xlabel("shard (dans l'ordre des clés)")
axes[0].set_ylabel("clés")
finish(fig, "Après une purge, la fusion évite une poussière de shards presque vides")
plt.show()
```

*2.* Sans fusion, les shards de la zone purgée restent en place avec quelques
dizaines de clés chacun : autant de surcoûts fixes pour presque rien. Avec la fusion,
ils se regroupent et la zone purgée retrouve des shards de taille raisonnable.

*3.* Prenons un shard plein (1 000 clés) et alternons l'insertion puis la
suppression d'une même clé. L'insertion provoque un débordement : scission en deux moitiés de
500 et 501 clés. Si `min_keys` = 600, la suppression suivante fait passer une moitié sous
le seuil, et la fusion (1 000 clés, sous la borne) est acceptée : on est revenu au point
de départ, après avoir réécrit 2 000 clés. Avec `min_keys` = 250, les deux moitiés
restent au-dessus du seuil et rien ne se passe.

```python
rows = []
for mn in [250, 450, 600]:
    d = MergingSharding(1_000, mn)
    for k in range(1_000):
        d.insert(2 * k)            # un shard plein (clés paires)
    base = d.rewritten
    for _ in range(200):
        d.insert(1)                # clé impaire : provoque la scission
        d.delete(1)                # la retire aussitôt
    rows.append({"min_keys": mn, "scissions": d.n_splits, "fusions": d.merges,
                 "clés réécrites pour 400 écritures": d.rewritten - base})
pd.DataFrame(rows).set_index("min_keys")
```

Avec `min_keys` = 600, chacune des 200 paires insertion/suppression déclenche une
scission et une fusion : des centaines de milliers de clés réécrites pour 400
écritures. Avec 250 ou 450, une seule scission a lieu, puis le système est stable. La
règle est celle de toute régulation à seuils : une **hystérésis**. Après une scission,
chaque moitié contient environ `max_keys / 2` clés ; il faut donc
`min_keys < max_keys / 2`, avec de la marge (un quart est un choix courant), pour qu'une
moitié fraîche ne soit pas aussitôt candidate à la fusion.

> **À retenir.** Le partitionnement dynamique doit savoir fusionner après des
> suppressions massives, sinon les shards vides s'accumulent. Les seuils de scission
> et de fusion doivent être nettement séparés (`min < max / 2`), sans quoi le système
> oscille et réécrit sans fin les mêmes données.

## Exercice 7.6 : éclater une clé chaude par salage

**Énoncé.** Un message devenu viral reçoit **40 000 écritures/s** (« j'aime » et
commentaires) et **300 000 lectures/s** de son compteur. Le cluster compte 32
shards ; chacun supporte 20 000 opérations/s et en sert déjà 8 000 en temps normal,
soit une marge de 12 000 opérations/s.

1. Pourquoi la clé ne tient-elle pas sur un shard ? Combien de suffixes $k$
   faudrait-il au minimum si les sous-clés tombaient toutes sur des shards distincts ?
2. Les sous-clés `post:7#00` ... sont hachées : deux d'entre elles peuvent tomber sur
   le même shard. Calculer la probabilité qu'aucun shard ne dépasse sa marge pour
   $k = 4$ à 8 et $k = 10$. Pourquoi n'est-elle pas croissante en $k$ ?
3. Une lecture du compteur doit lire les $k$ sous-clés et les additionner. Combien de
   lectures de sous-clés par seconde cela représente-t-il ? Proposer une solution pour
   les lectures.

**Rappels.** Le salage ajoute un suffixe aléatoire à une clé chaude pour étaler ses
écritures ; en contrepartie, chaque lecture doit interroger toutes les sous-clés, et la
charge de lecture par shard n'est pas réduite (cours 3, section *Skewed Workloads
and Relieving Hot Spots*).

**Solution.**

*1.* 40 000 écritures/s contre une marge de 12 000 : il faut au moins $k = 4$ sous-clés
(10 000 écritures/s chacune) si elles sont sur quatre shards distincts.

*2.* Chaque sous-clé porte $40\,000/k$ écritures/s ; un shard qui en reçoit $j$
dépasse sa marge si $j \cdot 40\,000/k > 12\,000$. Pour $k = 4$ (10 000 par sous-clé), il
faut $j \le 1$ : aucune collision, probabilité
$\frac{31}{32}\cdot\frac{30}{32}\cdot\frac{29}{32} \approx 0{,}82$. Pour $k = 5$ (8 000) et
$k = 6$ (6 667), deux sous-clés ensemble dépassent déjà la marge : il faut encore zéro
collision, ce qui devient **moins** probable quand $k$ augmente (0,72 puis 0,61). Pour
$k = 7$ (5 714) et $k = 8$ (5 000), on tolère deux sous-clés par shard (pas trois) ;
pour $k = 10$ (4 000), trois. La probabilité de succès remonte alors vers 1. Elle est en dents de
scie parce que le nombre de collisions tolérées $\lfloor 12\,000 \cdot k / 40\,000
\rfloor$ augmente par paliers.

*3.* $300\,000 \times k$ lectures de sous-clés par seconde, soit 3 millions pour
$k = 10$ ; et chaque shard hébergeant une sous-clé reçoit toujours 300 000
lectures/s, quinze fois sa capacité. Le salage règle les écritures, **pas** les lectures.
Pour les lectures, on sert un agrégat : un processus relit les $k$ sous-clés toutes les
100 ms et publie la somme dans un cache répliqué (10 × k lectures/s seulement sur le
stockage), quitte à afficher un compteur en retard de quelques dixièmes de seconde.

```python
W, S, MARGE, NPOSTS = 40_000, 32, 12_000, 2_000

def exacte(k):
    """Probabilité exacte qu'aucun shard ne dépasse la marge (dénombrement par récurrence)."""
    jmax = int(MARGE * k // W)
    # nombre de façons de placer k sous-clés numérotées dans S shards avec au plus jmax par shard
    from functools import lru_cache
    @lru_cache(None)
    def f(restantes, parts):
        if restantes == 0:
            return 1.0
        if parts == 0:
            return 0.0
        tot = 0.0
        for j in range(0, min(jmax, restantes) + 1):
            tot += math.comb(restantes, j) * f(restantes - j, parts - 1)
        return tot
    return f(k, S) / S ** k

rows = []
for k in [1, 2, 3, 4, 5, 6, 7, 8, 10, 12, 16]:
    pire = []
    for post in range(NPOSTS):
        sh = [P.stable_hash(v) % S for v in P.salted_variants(f"post:{post}", k)]
        pire.append(np.bincount(sh, minlength=S).max() * W / k)
    pire = np.array(pire)
    rows.append({"k": k, "P(succès) exacte": exacte(k), "P(succès) simulée": np.mean(pire <= MARGE),
                 "pire shard, médiane (écr/s)": np.median(pire),
                 "lectures de sous-clés/s (naïf)": 300_000 * k, "lectures/s (agrégat 100 ms)": 10 * k})
tab = pd.DataFrame(rows).set_index("k")

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(tab.index, tab.iloc[:, 0], "o-", color=INDIGO, label="probabilité exacte")
ax.plot(tab.index, tab.iloc[:, 1], "x", color=ROSE, ms=8, label=f"simulation ({NPOSTS} messages)")
for k, off in [(4, (-38, -4)), (6, (8, -6)), (7, (-38, -2)), (8, (6, -14))]:
    ax.annotate(f"k = {k}", (k, tab.loc[k].iloc[0]), off, textcoords="offset points", fontsize=8.5, color=SLATE)
ax.set_xlabel("nombre de sous-clés k"); ax.set_ylabel("P(aucun shard saturé)")
ax.set_ylim(-0.05, 1.05)
ax.set_title("Plus de sous-clés n'est pas toujours mieux : les collisions comptent")
subtitle(ax, "40 000 écritures/s réparties sur k sous-clés hachées dans 32 shards de marge 12 000/s")
ax.legend()
plt.tight_layout(); plt.show()
tab
```

Le dénombrement exact et la simulation sur 2 000 messages concordent : 0,82 pour $k = 4$,
puis une baisse à 0,72 et 0,61 pour $k = 5$ et 6, un saut à 0,97 pour $k = 7$ (deux
sous-clés tolérées par shard), un léger recul à $k = 8$, puis 0,99 et plus à partir de
$k = 10$. Les dents de scie se poursuivent au-delà, mais au-dessus de 0,95. En pratique, on prend une marge confortable ($k = 10$ à 16) ; la
dernière colonne rappelle que, côté lectures, seule une agrégation périodique rend le
coût indépendant du débit de lecture.

> **À retenir.** Le salage divise la charge d'**écriture** d'une clé chaude par $k$,
> mais multiplie les lectures par $k$ sans soulager le shard de chaque sous-clé.
> On le réserve à quelques clés connues, on choisit $k$ en tenant compte des collisions
> du hachage, et on sert les lectures par un agrégat mis en cache.
